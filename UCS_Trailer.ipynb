```python
import heapq

def uniform_cost_search(graph, start, goal):
    if start not in graph or goal not in graph:
        return None, []

    frontier = [(0, start, [start])]
    explored = set()

    while frontier:
        cost, node, path = heapq.heappop(frontier)

        print(f"Popped: {node} (cost {cost})")

        if node == goal:
            print(f"Goal reached: {' -> '.join(path)} ({cost} km)")
            return cost, path

        if node in explored:
            print(f"  Skipping {node}: already explored")
            continue
        explored.add(node)

        for neighbor, weight in graph[node].items():
            if neighbor not in explored:
                heapq.heappush(frontier, (cost + weight, neighbor, path + [neighbor]))
                print(f"  Queued: {neighbor} (cost {cost + weight})")

    return None, []


town_graph = {
    "nairobi": {"limuru": 38},
    "limuru": {"mai_mahiu": 20, "naivasha": 55},
    "mai_mahiu": {"naivasha": 35},
    "naivasha": {"nakuru": 71},
    "nakuru": {"eldama_ravine": 64, "kericho": 115, "eldoret": 156},
    "eldama_ravine": {"eldoret": 108},
    "kericho": {"awasi": 43},
    "eldoret": {"kapsabet": 46},
    "kapsabet": {"awasi": 56},
    "awasi": {"kisumu": 41},
    "kisumu": {},
}

cost, path = uniform_cost_search(town_graph, "nairobi", "kisumu")
if path:
    print(f"Shortest distance: {cost} km")
    print("Route:", " -> ".join(path))
else:
    print("No route found.")
```
